```python
from base64 import b64encode
from html import escape
from pathlib import Path
import csv
import os
import re
import shutil
import sys
import matplotlib.pyplot as plt
import numpy as np
import torch
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.patches import FancyArrowPatch
from matplotlib.lines import Line2D
from matplotlib.ticker import MaxNLocator
from PIL import Image
from pypdf import PdfReader, PdfWriter, Transformation
# Import shared Fig.02 style
sys.path.insert(0, str(PACKAGE_ROOT / "shared"))
from tool import apply_style
# ----------------------------- helper utilities ------------------------------
def _read_learning_csv(path: Path):
    with path.open(newline="") as f:
        rows = list(csv.reader(f))
    data = np.array(rows)
    epoch = data[1:, 0]
    train_loss = data[1:, 1].astype(np.float64)
    test_loss = data[1:, 2].astype(np.float64)
    loss_gap = test_loss - train_loss
    return epoch, train_loss, test_loss, loss_gap
def _load_sample_interactions(sample_dir: Path, salient_threshold: float):
    single_I_and = np.load(sample_dir / "after_sparsify" / "Iand.npy")
    single_I_or = np.load(sample_dir / "after_sparsify" / "Ior.npy")
    masks = np.load(sample_dir / "before_sparsify" / "masks.npy")
    single_I_and[np.abs(single_I_and) < salient_threshold] = 0
    single_I_or[np.abs(single_I_or) < salient_threshold] = 0
    return single_I_and, single_I_or, masks
# ----------------------------- disentangling math -----------------------------
def _generate_all_masks(length: int):
    masks = list(range(2**length))
    masks = [np.binary_repr(mask, width=length) for mask in masks]
    masks = [[bool(int(item)) for item in mask] for mask in masks]
    return masks
def _generate_subset_masks(set_mask, all_masks):
    set_mask_ = set_mask.expand_as(all_masks)
    is_subset = torch.logical_or(set_mask_, torch.logical_not(all_masks))
    is_subset = torch.all(is_subset, dim=1)
    return all_masks[is_subset], is_subset
def _triggering_mat_and(dim: int):
    all_masks = torch.BoolTensor(_generate_all_masks(dim))
    n_masks, _ = all_masks.shape
    mat = []
    for i in range(n_masks):
        mask_S = all_masks[i]
        row = torch.zeros(n_masks)
        mask_Ls, L_indices = _generate_subset_masks(mask_S, all_masks)
        row[L_indices] = 1.0
        mat.append(row.clone())
    return torch.stack(mat).double()
def _M_mat_and(dim: int, sigma2: float):
    J = _triggering_mat_and(dim)
    JtJ = torch.mm(J.t(), J)
    masks = torch.BoolTensor(_generate_all_masks(dim))
    masks_sum = torch.sum(masks, dim=1, dtype=torch.float64)
    c = torch.pow(2, masks_sum) * sigma2
    diag_c = torch.diag(c) * 2**dim
    return torch.linalg.inv(JtJ + diag_c) @ JtJ
def _generate_set_with_intersection_masks(set_mask, all_masks):
    set_mask_ = set_mask.expand_as(all_masks)
    have_intersection = torch.logical_and(set_mask_, all_masks)
    have_intersection = torch.any(have_intersection, dim=1)
    return all_masks[have_intersection], have_intersection
def _triggering_mat_or(dim: int):
    all_masks = torch.BoolTensor(_generate_all_masks(dim))
    n_masks, _ = all_masks.shape
    mat = []
    mask_empty = torch.zeros(dim).bool()
    _, empty_indice = _generate_subset_masks(mask_empty, all_masks)
    for i in range(n_masks):
        mask_S = all_masks[i]
        row = torch.zeros(n_masks)
        row[empty_indice] = 1.0
        mask_Ls, L_indices = _generate_set_with_intersection_masks(mask_S, all_masks)
        row[L_indices] = 1.0
        mat.append(row.clone())
    return torch.stack(mat).double()
def _M_mat_or(dim: int, sigma2: float):
    J = _triggering_mat_or(dim)
    JtJ = torch.mm(J.t(), J)
    masks = torch.BoolTensor(_generate_all_masks(dim))
    masks_sum = torch.sum(masks, dim=1, dtype=torch.float64)
    c = torch.pow(2, masks_sum) * sigma2
    diag_c = torch.diag(c) * 2**dim
    return torch.linalg.inv(JtJ + diag_c) @ JtJ
def _general_comb(n, k):
    if n < k:
        return torch.tensor(0.0, dtype=torch.float64)
    return torch.math.gamma(n + 1) / (torch.math.gamma(k + 1) * torch.math.gamma(n - k + 1))
def _compute_noise_interaction_distribution(alpha, beta, order=8):
    alpha_order = order * alpha
    general_comb_value = torch.tensor([_general_comb(alpha_order, j) for j in range(1, order + 1)])
    interaction = general_comb_value * beta
    return interaction
def _compute_reliable_interaction_distribution(sigma2, gamma, w_star_and, w_star_or, dim=8, tau=TAU):
    M_and = _M_mat_and(dim, sigma2)
    w_hat_and = M_and @ w_star_and * gamma
    M_or = _M_mat_or(dim, sigma2)
    w_hat_or = M_or @ w_star_or * gamma
    salient_threshold = torch.abs(torch.sum(w_hat_and[1:]) + torch.sum(w_hat_or[1:])) * tau
    w_hat_and[torch.abs(w_hat_and) < salient_threshold] = 0.0
    w_hat_or[torch.abs(w_hat_or) < salient_threshold] = 0.0
    return w_hat_and, w_hat_or
# ----------------------------- plotting core ---------------------------------
def disentangling_contributions(
    single_I_and_list,
    single_star_I_and_list,
    single_I_or_list,
    single_star_I_or_list,
    masks,
    ax,
    cache_file: Path,
    fig_i: int = 0,
    row_colors=None,
    tau: float = TAU,
):
    """Plot a single column of the 4x4 disentanglement figure.
    Returns
    -------
    alpha, beta, sigma2, gamma, norm_reliable, norm_noise, norm_res
    """
    if row_colors is None:
        row_colors = [
            ("#DE582B", "#1868B2"),
            ("#F3A332", "#018A67"),
            ("#F3A332", "#018A67"),
            ("#F3A332", "#018A67"),
        ]
    cache_file = Path(cache_file)
    cache_file.parent.mkdir(parents=True, exist_ok=True)
    n_dim = masks.shape[1]
    i_orders = np.sum(masks, axis=1).astype(int)
    num_sample = len(single_I_and_list)
    opt_cache = cache_file.parent / f"{cache_file.name}_fig{fig_i}_tau={tau:.4f}_opt.pt"
    single_I_and_list = torch.tensor(single_I_and_list, dtype=torch.float64)
    single_star_I_and_list = torch.tensor(single_star_I_and_list, dtype=torch.float64)
    single_I_or_list = torch.tensor(single_I_or_list, dtype=torch.float64)
    single_star_I_or_list = torch.tensor(single_star_I_or_list, dtype=torch.float64)
    # Normalize w.r.t. order-1 interactions
    indices = i_orders == 1
    single_I_pos_order1 = (
        torch.where(single_I_and_list[:, indices] > 0, single_I_and_list[:, indices], torch.tensor(0.0))
        + torch.where(single_I_or_list[:, indices] > 0, single_I_or_list[:, indices], torch.tensor(0.0))
    )
    single_I_neg_order1 = (
        -torch.where(single_I_and_list[:, indices] < 0, single_I_and_list[:, indices], torch.tensor(0.0))
        - torch.where(single_I_or_list[:, indices] < 0, single_I_or_list[:, indices], torch.tensor(0.0))
    )
    normalize_factor = torch.linalg.norm(single_I_pos_order1, ord=1) + torch.linalg.norm(single_I_neg_order1, ord=1)
    normalize_factor = normalize_factor.item() / num_sample
    single_I_and_list = single_I_and_list / normalize_factor
    single_I_or_list = single_I_or_list / normalize_factor
    single_I_pos_order1 = (
        torch.where(single_star_I_and_list > 0, single_star_I_and_list, torch.tensor(0.0))
        + torch.where(single_star_I_or_list > 0, single_star_I_or_list, torch.tensor(0.0))
    )
    single_I_neg_order1 = (
        -torch.where(single_star_I_and_list < 0, single_star_I_and_list, torch.tensor(0.0))
        - torch.where(single_star_I_or_list < 0, single_star_I_or_list, torch.tensor(0.0))
    )
    normalize_factor = torch.linalg.norm(single_I_pos_order1, ord=1) + torch.linalg.norm(single_I_neg_order1, ord=1)
    normalize_factor = normalize_factor.item() / num_sample
    single_star_I_and_list = single_star_I_and_list / normalize_factor
    single_star_I_or_list = single_star_I_or_list / normalize_factor
    pos_single_I_and = torch.where(single_I_and_list > 0, single_I_and_list, torch.tensor(0.0))
    neg_single_I_and = -torch.where(single_I_and_list < 0, single_I_and_list, torch.tensor(0.0))
    pos_single_I_or = torch.where(single_I_or_list > 0, single_I_or_list, torch.tensor(0.0))
    neg_single_I_or = -torch.where(single_I_or_list < 0, single_I_or_list, torch.tensor(0.0))
    pos_single_I_each_order = torch.zeros(n_dim)
    neg_single_I_each_order = torch.zeros(n_dim)
    for i_order in range(1, n_dim + 1):
        indices = i_orders == i_order
        pos_single_I_each_order[i_order - 1] = (
            pos_single_I_and[:, indices].sum() + pos_single_I_or[:, indices].sum()
        ) / num_sample
        neg_single_I_each_order[i_order - 1] = (
            neg_single_I_and[:, indices].sum() + neg_single_I_or[:, indices].sum()
        ) / num_sample
    target_interaction_pos = torch.clone(pos_single_I_each_order)
    target_interaction_neg = torch.clone(neg_single_I_each_order)
    opt = None
    if opt_cache.exists():
        loaded_opt = torch.load(opt_cache, weights_only=True)
        has_matching_target = (
            "target_interaction_pos" in loaded_opt
            and "target_interaction_neg" in loaded_opt
            and torch.allclose(loaded_opt["target_interaction_pos"], target_interaction_pos)
            and torch.allclose(loaded_opt["target_interaction_neg"], target_interaction_neg)
        )
        if has_matching_target:
            opt = loaded_opt
    if opt is not None:
        alpha = opt["alpha"]
        beta = opt["beta"]
        sigma2 = opt["sigma2"]
        gamma = opt["gamma"]
        reliable_interation_pos = opt["reliable_interation_pos"]
        reliable_interation_neg = opt["reliable_interation_neg"]
        noise_interaction = opt["noise_interaction"]
        pos_res_each_order = opt["pos_res_each_order"]
        neg_res_each_order = opt["neg_res_each_order"]
        norm_reliable = opt["norm_reliable"]
        norm_noise = opt["norm_noise"]
        norm_res = opt["norm_res"]
    else:
        alpha_list = torch.linspace(0, 300, 101) * 1e-2
        beta_list = torch.linspace(0, 5000, 1001) * 1e-3
        # Noise metric grid
        noise_interaction_metric = torch.zeros(alpha_list.shape[0], beta_list.shape[0], n_dim)
        for i, alpha_item in enumerate(alpha_list):
            noise_interaction_metric_1 = _compute_noise_interaction_distribution(alpha_item, 1, order=n_dim)
            beta_noise = noise_interaction_metric_1.unsqueeze(0).repeat(beta_list.shape[0], 1) * beta_list.unsqueeze(1).repeat(1, n_dim)
            noise_interaction_metric[i] = beta_noise
        noise_interaction_metric = noise_interaction_metric.unsqueeze(2).unsqueeze(2)
        sigma2_list = torch.tensor([1e-2, 1e-1], dtype=torch.float64)
        gamma_list = torch.linspace(-1, 4, 501) + 1
        # Reliable interaction grid
        reliable_interation_pos_metric = torch.zeros(sigma2_list.shape[0], gamma_list.shape[0], n_dim)
        reliable_interation_neg_metric = torch.zeros(sigma2_list.shape[0], gamma_list.shape[0], n_dim)
        reliable_orders_by_sigma = {}
        reliable_interation_pos = torch.zeros(2**n_dim)
        reliable_interation_neg = torch.zeros(2**n_dim)
        for sigma2_i, sigma2_item in enumerate(sigma2_list):
            cache_pos = cache_file.parent / f"{cache_file.name}_pos_{sigma2_item.item()}.out"
            cache_neg = cache_file.parent / f"{cache_file.name}_neg_{sigma2_item.item()}.out"
            if cache_pos.exists() and cache_neg.exists():
                reliable_interation_pos_each_order = torch.load(cache_pos, weights_only=True)
                reliable_interation_neg_each_order = torch.load(cache_neg, weights_only=True)
            else:
                reliable_interation_pos.zero_()
                reliable_interation_neg.zero_()
                for i in range(num_sample):
                    star_I_and = single_star_I_and_list[i]
                    star_I_or = single_star_I_or_list[i]
                    reliable_I_and, reliable_I_or = _compute_reliable_interaction_distribution(
                        sigma2_item, torch.tensor(1.0, dtype=torch.float64), star_I_and, star_I_or, dim=n_dim, tau=tau
                    )
                    reliable_pos = torch.where(reliable_I_and > 0, reliable_I_and, torch.tensor(0.0)) + torch.where(reliable_I_or > 0, reliable_I_or, torch.tensor(0.0))
                    reliable_neg = -torch.where(reliable_I_and < 0, reliable_I_and, torch.tensor(0.0)) - torch.where(reliable_I_or < 0, reliable_I_or, torch.tensor(0.0))
                    reliable_interation_pos += reliable_pos
                    reliable_interation_neg += reliable_neg
                reliable_interation_pos /= num_sample
                reliable_interation_neg /= num_sample
                reliable_interation_pos_each_order = torch.zeros(n_dim)
                reliable_interation_neg_each_order = torch.zeros(n_dim)
                for i_order in range(1, n_dim + 1):
                    indices = i_orders == i_order
                    reliable_interation_pos_each_order[i_order - 1] = reliable_interation_pos[indices].sum()
                    reliable_interation_neg_each_order[i_order - 1] = reliable_interation_neg[indices].sum()
                torch.save(reliable_interation_pos_each_order, cache_pos)
                torch.save(reliable_interation_neg_each_order, cache_neg)
            reliable_orders_by_sigma[float(sigma2_item)] = (reliable_interation_pos_each_order, reliable_interation_neg_each_order)
            gamma_reliable_pos = reliable_interation_pos_each_order.unsqueeze(0).repeat(gamma_list.shape[0], 1) * gamma_list.unsqueeze(1).repeat(1, n_dim)
            gamma_reliable_neg = reliable_interation_neg_each_order.unsqueeze(0).repeat(gamma_list.shape[0], 1) * gamma_list.unsqueeze(1).repeat(1, n_dim)
            reliable_interation_pos_metric[sigma2_i] = gamma_reliable_pos
            reliable_interation_neg_metric[sigma2_i] = gamma_reliable_neg
        reliable_interation_pos_metric = reliable_interation_pos_metric.unsqueeze(0).unsqueeze(0)
        reliable_interation_neg_metric = reliable_interation_neg_metric.unsqueeze(0).unsqueeze(0)
        all_interation_pos_metric = reliable_interation_pos_metric + noise_interaction_metric
        all_interation_neg_metric = reliable_interation_neg_metric + noise_interaction_metric
        loss_metric = torch.linalg.norm(all_interation_pos_metric - target_interaction_pos, dim=-1, ord=2) + torch.linalg.norm(all_interation_neg_metric - target_interaction_neg, dim=-1, ord=2)
        min_loss = loss_metric.min()
        min_loss_index = torch.where(loss_metric == min_loss)
        min_loss_alpha_index = min_loss_index[0][0]
        min_loss_beta_index = min_loss_index[1][0]
        min_loss_sigma2_index = min_loss_index[2][0]
        min_loss_gamma_index = min_loss_index[3][0]
        alpha = alpha_list[min_loss_alpha_index]
        beta = beta_list[min_loss_beta_index]
        sigma2 = sigma2_list[min_loss_sigma2_index]
        gamma = gamma_list[min_loss_gamma_index]
        noise_interaction = _compute_noise_interaction_distribution(alpha, beta, order=n_dim)
        reliable_interation_pos, reliable_interation_neg = reliable_orders_by_sigma[float(sigma2)]
        reliable_interation_pos = reliable_interation_pos * gamma
        reliable_interation_neg = reliable_interation_neg * gamma
        pos_res_each_order = target_interaction_pos - (reliable_interation_pos + noise_interaction)
        neg_res_each_order = target_interaction_neg - (reliable_interation_neg + noise_interaction)
        norm_reliable = torch.linalg.norm(reliable_interation_pos, ord=1) + torch.linalg.norm(reliable_interation_neg, ord=1)
        norm_noise = 2 * torch.linalg.norm(noise_interaction, ord=1)
        norm_res = torch.linalg.norm(pos_res_each_order, ord=1) + torch.linalg.norm(neg_res_each_order, ord=1)
        torch.save(
            {
                "alpha": alpha,
                "beta": beta,
                "sigma2": sigma2,
                "gamma": gamma,
                "target_interaction_pos": target_interaction_pos,
                "target_interaction_neg": target_interaction_neg,
                "reliable_interation_pos": reliable_interation_pos,
                "reliable_interation_neg": reliable_interation_neg,
                "noise_interaction": noise_interaction,
                "pos_res_each_order": pos_res_each_order,
                "neg_res_each_order": neg_res_each_order,
                "norm_reliable": norm_reliable,
                "norm_noise": norm_noise,
                "norm_res": norm_res,
            },
            opt_cache,
        )
    X = np.arange(n_dim)
    values = [
        (target_interaction_pos, target_interaction_neg),
        (reliable_interation_pos, reliable_interation_neg),
        (noise_interaction, noise_interaction),
        (torch.abs(pos_res_each_order), torch.abs(neg_res_each_order)),
    ]
    for i, (pos_vals, neg_vals) in enumerate(values):
        pos_color, neg_color = row_colors[i]
        ax[i].bar(X, pos_vals, color=pos_color, edgecolor="none")
        ax[i].bar(X, -neg_vals, color=neg_color, edgecolor="none")
        # X-axis: show ticks and label in every subplot
        ax[i].set_xticks([0, n_dim - 1])
        ax[i].set_xticklabels(["1", str(n_dim)], fontsize=FONT_SIZE_TICK_LABEL - 1.5, color="black")
        ax[i].set_xlabel("Order $m$", fontsize=FONT_SIZE_TICK_LABEL - 1.5, color="black")
        ax[i].xaxis.set_label_coords(0.5, -0.05)
        # Y-axis: show tick labels in every subplot
        ax[i].tick_params(axis="both", labelsize=FONT_SIZE_TICK_LABEL - 1.5)
        ax[i].grid(False)
        ax[i].spines["top"].set_visible(False)
        ax[i].spines["right"].set_visible(False)
        ax[i].spines["left"].set_linewidth(0.6)
        ax[i].spines["bottom"].set_linewidth(0.6)
    return alpha.item(), beta.item(), sigma2.item(), gamma.item(), norm_reliable.item(), norm_noise.item(), norm_res.item()
def _parse_epoch_label(label: str):
    """Extract numeric epoch from label like epoch_0, iter_57, epoch_256.
    
    iter_* labels are treated as epoch 1.
    Returns (display_number, is_iter).
    """
    if label.startswith("iter_"):
        return 1, True
    elif label.startswith("epoch_"):
        try:
            return int(label.split("_")[1]), False
        except (ValueError, IndexError):
            return 0, False
    else:
        try:
            return int(label), False
        except ValueError:
            return 0, False
def plot_loss_and_gap(
    epoch,
    train_loss,
    test_loss,
    loss_gap,
    loss_choose_index,
    interact_choose_index,
    interact_choose_index_in_loss_index,
    ax,
    show_xaxis: bool = True,
    clip_epoch0_loss: bool = False,
    legend_ax=None,
    show_legend: bool = True,
):
    """Plot train/test loss and loss gap as a training-stage scaffold.
    The four selected timepoints are mapped to x = 0.5, 1.5, 2.5, 3.5 so that
    all timepoint markers align with the centers of the distribution columns.
    """
    # Build filtered arrays: keep only epoch_* entries
    kept_positions = []
    epoch_numbers = []
    for i, idx in enumerate(loss_choose_index):
        label = str(epoch[idx])
        num, is_iter = _parse_epoch_label(label)
        if not is_iter:
            kept_positions.append(i)
            epoch_numbers.append(num)
    n_kept = len(kept_positions)
    chosen_train = train_loss[loss_choose_index][kept_positions]
    chosen_test = test_loss[loss_choose_index][kept_positions]
    chosen_gap = loss_gap[loss_choose_index][kept_positions]
    # Remap each timepoint to its position in the filtered array
    remapped_timepoints = []
    for orig_idx in interact_choose_index_in_loss_index:
        label = str(epoch[loss_choose_index[orig_idx]])
        _, is_iter = _parse_epoch_label(label)
        if is_iter:
            try:
                remapped_timepoints.append(epoch_numbers.index(1))
            except ValueError:
                remapped_timepoints.append(0)
        else:
            remapped_timepoints.append(kept_positions.index(orig_idx))
    # Scale x to [0, 4] with timepoints at the centers of the four columns
    target_tp_x = np.array([0.5, 1.5, 2.5, 3.5])
    x_scaled = np.zeros(n_kept)
    for i in range(n_kept):
        if i <= remapped_timepoints[0]:
            denom = remapped_timepoints[0]
            x_scaled[i] = target_tp_x[0] * (i / denom) if denom > 0 else target_tp_x[0]
        elif i >= remapped_timepoints[-1]:
            n_after = n_kept - 1 - remapped_timepoints[-1]
            x_scaled[i] = target_tp_x[-1] + (4.0 - target_tp_x[-1]) * ((i - remapped_timepoints[-1]) / n_after) if n_after > 0 else target_tp_x[-1]
        else:
            for j in range(len(remapped_timepoints) - 1):
                t0 = remapped_timepoints[j]
                t1 = remapped_timepoints[j + 1]
                if t0 <= i <= t1:
                    t = (i - t0) / (t1 - t0) if t1 > t0 else 0.0
                    x_scaled[i] = target_tp_x[j] + t * (target_tp_x[j + 1] - target_tp_x[j])
                    break
    ax.set_xlim(0, 4)
    if clip_epoch0_loss:
        lower_min, lower_max = -0.05, 0.55
        upper_min = min(0.90, float(min(chosen_train[0], chosen_test[0])) - 0.05)
        upper_max = max(1.25, float(max(chosen_train[0], chosen_test[0])) + 0.05)
        lower_height = 0.64
        break_gap = 0.10
        upper_height = 0.26
        def map_loss_y(values):
            values = np.asarray(values, dtype=float)
            mapped = np.empty_like(values, dtype=float)
            lower_mask = values <= lower_max
            upper_mask = values >= upper_min
            middle_mask = ~(lower_mask | upper_mask)
            mapped[lower_mask] = lower_height * (values[lower_mask] - lower_min) / (lower_max - lower_min)
            mapped[upper_mask] = lower_height + break_gap + upper_height * (values[upper_mask] - upper_min) / (upper_max - upper_min)
            mapped[middle_mask] = lower_height + break_gap * 0.5
            return mapped
        chosen_train_plot = map_loss_y(chosen_train)
        chosen_test_plot = map_loss_y(chosen_test)
        chosen_gap_plot = map_loss_y(chosen_gap)
        loss_top = lower_height + break_gap + upper_height
        ymin = 0.0
        display_span = 1.0
        train_clipped = np.zeros_like(chosen_train, dtype=bool)
        test_clipped = np.zeros_like(chosen_test, dtype=bool)
        broken_axis_ticks = np.array([0.0, 0.25, 0.50, 1.0, 1.25])
        broken_axis_tick_positions = map_loss_y(broken_axis_ticks)
        broken_axis_tick_labels = [f"{tick:g}" for tick in broken_axis_ticks]
    else:
        data_min = min(float(np.min(chosen_train)), float(np.min(chosen_test)), float(np.min(chosen_gap)))
        data_max = max(float(np.max(chosen_train)), float(np.max(chosen_test)), float(np.max(chosen_gap)))
        data_span = max(data_max - data_min, 1.0)
        ymin = min(-0.15, data_min - 0.08 * data_span)
        loss_top = data_max
        display_span = max(loss_top - ymin, 1.0)
        train_clipped = np.zeros_like(chosen_train, dtype=bool)
        test_clipped = np.zeros_like(chosen_test, dtype=bool)
        chosen_train_plot = chosen_train
        chosen_test_plot = chosen_test
        chosen_gap_plot = chosen_gap
        broken_axis_tick_positions = None
        broken_axis_tick_labels = None
    # Reserve an annotation band above the plotted loss curves.
    if clip_epoch0_loss:
        ymax = 1.85
    else:
        annotation_band = 0.50 * data_span
        ymax = data_max + annotation_band
    ax.set_ylim(ymin, ymax)
    phase_boundary = target_tp_x[1]
    # Vertical positions of the three annotation layers (axes coordinates).
    PHASE_TEXT_Y = 1.08
    ARROW_Y = 0.78
    ax.axvspan(0, phase_boundary, color="#F4F0FA", alpha=0.22, zorder=0)
    ax.axvspan(phase_boundary, 4, color="#F1F7EF", alpha=0.22, zorder=0)
    if clip_epoch0_loss:
        ax.axhspan(lower_height, lower_height + break_gap, color="white", alpha=1.0, zorder=1.5)
        ax.plot([0.0, 0.0], [lower_height, lower_height + break_gap], color="white", lw=2.2, zorder=8, clip_on=False)
    # Vertical guide lines at column centers, aligned with the distribution panels below.
    for c, tp_x in enumerate(target_tp_x):
        if c == 1:
            ax.axvline(tp_x, color="#777777", linestyle="--", linewidth=1.0, alpha=0.70, zorder=1)
        else:
            ax.axvline(tp_x, color="#B8B8B8", linestyle="--", linewidth=0.8, alpha=0.55, zorder=1)
    # Train / test loss curves.
    train_line, = ax.plot(
        x_scaled, chosen_train_plot, label="Training loss", color=COLOR_TRAIN,
        linewidth=1.5, marker="o", markersize=2.6,
        markerfacecolor=COLOR_TRAIN, markeredgecolor="white", markeredgewidth=0.5,
        alpha=0.82,
        zorder=4,
    )
    test_line, = ax.plot(
        x_scaled, chosen_test_plot, label="Testing loss", color=COLOR_TEST,
        linewidth=1.5, marker="s", markersize=2.6,
        markerfacecolor=COLOR_TEST, markeredgecolor="white", markeredgewidth=0.5,
        alpha=0.82,
        zorder=4,
    )
    if clip_epoch0_loss:
        break_y0 = lower_height / ymax
        break_y1 = (lower_height + break_gap) / ymax
        slash_dx = 0.018
        slash_dy = 0.018
        for y_break in [break_y0, break_y1]:
            ax.plot([-slash_dx, slash_dx], [y_break - slash_dy, y_break + slash_dy], transform=ax.transAxes, color="#444444", lw=1.0, clip_on=False, solid_capstyle="butt")
            ax.plot([-slash_dx, slash_dx], [y_break - 0.002, y_break + 2 * slash_dy - 0.002], transform=ax.transAxes, color="#444444", lw=1.0, clip_on=False, solid_capstyle="butt")
    # Loss gap is the primary metric.
    ax.fill_between(
        x_scaled, chosen_train_plot, chosen_test_plot,
        color=COLOR_GAP, alpha=0.05,
        zorder=2,
    )
    gap_line, = ax.plot(
        x_scaled, chosen_gap_plot, label="Loss gap",
        color="#333333", linewidth=2.1, linestyle="-",
        marker="D", markersize=3.2,
        markerfacecolor="white", markeredgecolor="#333333", markeredgewidth=1.0,
        alpha=0.95,
        zorder=5,
    )
    # Phase scaffold: three clear vertical layers inside the loss panel.
    phase_lw = 2.4
    xaxis_tf = ax.get_xaxis_transform()
    ax.plot(
        [0.08, phase_boundary], [ARROW_Y, ARROW_Y],
        transform=xaxis_tf,
        color=PHASE_NORMAL_ARROW_END, linewidth=phase_lw,
        solid_capstyle="round", zorder=6,
    )
    ax.annotate(
        "",
        xy=(3.95, ARROW_Y), xytext=(phase_boundary, ARROW_Y),
        xycoords=xaxis_tf, textcoords=xaxis_tf,
        arrowprops=dict(arrowstyle="-|>", color=PHASE_OVERFIT_ARROW_END, lw=phase_lw, mutation_scale=8),
        zorder=6,
    )
    ax.text(
        phase_boundary / 2 + 0.10, PHASE_TEXT_Y, "Normal learning phase",
        transform=xaxis_tf,
        ha="center", va="top", fontsize=FONT_SIZE_PHASE_LABEL,
        color="#4F4F4F", fontweight="bold", zorder=8,
    )
    ax.text(
        (phase_boundary + target_tp_x[-1]) / 2 + 0.45, PHASE_TEXT_Y, "Over-fitting phase",
        transform=xaxis_tf,
        ha="center", va="top", fontsize=FONT_SIZE_PHASE_LABEL,
        color="#4F4F4F", fontweight="bold", zorder=8,
    )
    for c, tp_x in enumerate(target_tp_x):
        ax.plot(
            tp_x, ARROW_Y, "o",
            transform=xaxis_tf,
            markersize=4.5,
            markerfacecolor="white",
            markeredgecolor="#333333",
            markeredgewidth=1.2,
            zorder=9,
        )
    if show_xaxis:
        ax.set_xticks(target_tp_x)
        ax.set_xticklabels([epoch_numbers[i] for i in remapped_timepoints], fontsize=FONT_SIZE_TICK_LABEL)
        ax.set_xlabel("Training Epoch", fontsize=FONT_SIZE_AXIS_LABEL, labelpad=3)
    else:
        ax.set_xticks([])
        ax.set_xticklabels([])
        ax.set_xlabel("")
    ax.set_ylabel("Loss", fontsize=FONT_SIZE_ROW_LABEL + 2, fontweight="bold", labelpad=3)
    if broken_axis_tick_positions is not None:
        ax.set_yticks(broken_axis_tick_positions)
        ax.set_yticklabels(broken_axis_tick_labels)
    ax.tick_params(axis="y", labelsize=FONT_SIZE_TICK_LABEL)
    ax.grid(True, axis="y", alpha=0.18, linestyle="--", linewidth=0.4, color="#BBBBBB")
    ax.grid(False, axis="x")
    for spine in ax.spines.values():
        spine.set_linewidth(0.6)
    def choose_loss_legend_anchor():
        dense_x = np.linspace(1.7, float(x_scaled[-1]), 60)
        data_y_frac = []
        for series in [chosen_gap_plot, chosen_train_plot, chosen_test_plot]:
            dense_y = np.interp(dense_x, x_scaled, np.asarray(series, dtype=float))
            data_y_frac.extend(((dense_y - ymin) / (ymax - ymin)).tolist())
        annotation_y_frac = np.array([ARROW_Y, PHASE_TEXT_Y])
        candidates = [0.76, 0.74, 0.72, 0.70, 0.66, 0.58, 0.50, 0.42, 0.34, 0.26]
        legend_h = 0.105
        best_anchor = candidates[0]
        best_score = float("inf")
        for anchor_y in candidates:
            y0 = anchor_y - legend_h
            y1 = anchor_y + 0.015
            data_penalty = sum(1.0 for y in data_y_frac if y0 <= y <= y1)
            near_data_penalty = sum(max(0.0, 0.045 - min(abs(y - y0), abs(y - y1))) for y in data_y_frac)
            annotation_penalty = sum(3.0 for y in annotation_y_frac if y0 - 0.03 <= y <= y1 + 0.03)
            score = data_penalty + near_data_penalty + annotation_penalty + 0.01 * abs(anchor_y - 0.58)
            if score < best_score:
                best_score = score
                best_anchor = anchor_y
        return (0.98, best_anchor)
    legend_handles = [gap_line, train_line, test_line]
    if show_legend and legend_ax is not None:
        legend_ax.axis("off")
        legend_ax.legend(
            handles=legend_handles,
            loc="center", ncol=len(legend_handles),
            fontsize=FONT_SIZE_LEGEND - 1, frameon=False,
            borderpad=0.0, handletextpad=0.3,
            columnspacing=0.8, labelspacing=0.25,
            handlelength=1.2, markerscale=0.75,
        )
    elif show_legend:
        legend_anchor = choose_loss_legend_anchor()
        ax.legend(
            handles=legend_handles,
            loc="upper right", bbox_to_anchor=legend_anchor,
            fontsize=FONT_SIZE_LEGEND - 1, frameon=False,
            ncol=len(legend_handles), borderpad=0.2, handletextpad=0.3,
            columnspacing=0.8, labelspacing=0.25,
            handlelength=1.2, markerscale=0.75,
        )
    return ax, remapped_timepoints, epoch_numbers
# ----------------------------- model-level driver -----------------------------
def generate_model_figures(config: dict, output_root: Path, cache_root: Path, font_size: int = 8):
    """Generate loss_and_gap and disentangle PNG/PDF files for one model."""
    key = config["key"]
    out_dir = output_root / key
    out_dir.mkdir(parents=True, exist_ok=True)
    cache_dir = cache_root / key
    cache_dir.mkdir(parents=True, exist_ok=True)
    apply_style(font_size)
    epoch, train_loss, test_loss, loss_gap = _read_learning_csv(config["learning_csv"])
    # Loss and gap
    fig_loss, ax_loss = plt.subplots(figsize=LOSS_FIGSIZE)
    _, remapped_timepoints, epoch_numbers = plot_loss_and_gap(
        epoch,
        train_loss,
        test_loss,
        loss_gap,
        config["loss_choose_index"],
        config["interact_choose_index"],
        config["interact_choose_index_in_loss_index"],
        ax=ax_loss,
        clip_epoch0_loss=config.get("clip_epoch0_loss", False),
    )
    fig_loss.subplots_adjust(left=LOSS_LEFT, right=LOSS_RIGHT, top=LOSS_TOP, bottom=LOSS_BOTTOM)
    loss_path = out_dir / "loss_and_gap.png"
    fig_loss.savefig(loss_path, bbox_inches=None, pad_inches=0.0, dpi=300)
    fig_loss.savefig(out_dir / "loss_and_gap.pdf", bbox_inches=None, pad_inches=0.0)
    plt.close(fig_loss)
    timepoint_labels = config.get("timepoint_labels", TIMEPOINT_LABELS)
    if config.get("archived_dynamic", False):
        star_epoch = timepoint_labels[-1]
        star_interaction_dir = config["interaction_path"] / f"Harsanyi_{star_epoch}"
    else:
        # Star interaction reference at epoch 256
        star_epoch = 256
        star_interaction_dir = config["interaction_path"] / f"before_epoch_{star_epoch}"
    star_samples = sorted(os.listdir(star_interaction_dir))
    vn_v0_list = []
    for sample in star_samples:
        rewards = np.load(star_interaction_dir / sample / "before_sparsify" / "rewards.npy")
        vn_v0_list.append(np.abs(rewards[-1] - rewards[0]))
    salient_threshold_star = np.mean(vn_v0_list) * TAU
    star_I_and_list = []
    star_I_or_list = []
    masks = None
    for sample in star_samples:
        single_I_and, single_I_or, masks = _load_sample_interactions(star_interaction_dir / sample, salient_threshold_star)
        star_I_and_list.append(single_I_and)
        star_I_or_list.append(single_I_or)
    star_I_and_list = np.array(star_I_and_list)
    star_I_or_list = np.array(star_I_or_list)
    # 4x4 disentanglement grid with custom height ratios
    fig, axes = plt.subplots(
        4, 4, figsize=DIS_FIGSIZE, sharey=False, sharex="col", height_ratios=DIS_HEIGHT_RATIOS
    )
    fig.subplots_adjust(
        wspace=DIS_WSPACE,
        hspace=DIS_HSPACE,
        left=DIS_LEFT,
        right=DIS_RIGHT,
        top=DIS_TOP,
        bottom=DIS_BOTTOM,
    )
    row_colors = ROW_COLORS
    norm_reliable_list = []
    norm_noise_list = []
    norm_res_list = []
    for fig_i, index in enumerate(config["interact_choose_index"]):
        if config.get("archived_dynamic", False):
            interaction_dir = config["interaction_path"] / f"Harsanyi_{timepoint_labels[fig_i]}"
        else:
            interaction_dir = config["interaction_path"] / f"before_{epoch[index]}"
        samples = sorted(os.listdir(interaction_dir))
        vn_v0_list = []
        for sample in samples:
            rewards = np.load(interaction_dir / sample / "before_sparsify" / "rewards.npy")
            vn_v0_list.append(np.abs(rewards[-1] - rewards[0]))
        salient_threshold = np.mean(vn_v0_list) * TAU
        single_I_and_list = []
        single_I_or_list = []
        for sample in samples:
            single_I_and, single_I_or, _ = _load_sample_interactions(interaction_dir / sample, salient_threshold)
            single_I_and_list.append(single_I_and)
            single_I_or_list.append(single_I_or)
        single_I_and_list = np.array(single_I_and_list)
        single_I_or_list = np.array(single_I_or_list)
        ax_col = axes[:, fig_i]
        cache_file = cache_dir / f"star_{star_epoch}"
        alpha, beta, sigma, gamma, norm_reliable, norm_noise, norm_res = disentangling_contributions(
            single_I_and_list,
            star_I_and_list,
            single_I_or_list,
            star_I_or_list,
            masks,
            ax_col,
            cache_file=cache_file,
            fig_i=fig_i,
            row_colors=row_colors,
        )
        norm_sum = norm_reliable + norm_noise + norm_res
        norm_reliable_list.append(norm_reliable / norm_sum)
        norm_noise_list.append(norm_noise / norm_sum)
        norm_res_list.append(norm_res / norm_sum)
        print(
            f"[{key}] epoch={epoch[index]} alpha={alpha:.4f} beta={beta:.4f} "
            f"sigma={sigma:.4f} gamma={gamma:.4f} reliable={norm_reliable_list[-1]:.4f} "
            f"noise={norm_noise_list[-1]:.4f} res={norm_res_list[-1]:.4f}"
        )
    # Unify all rows to the first row's y-axis range
    ylim_mins = [axes[0, c].get_ylim()[0] for c in range(axes.shape[1])]
    ylim_maxs = [axes[0, c].get_ylim()[1] for c in range(axes.shape[1])]
    common_ylim = (min(ylim_mins), max(ylim_maxs))
    for ax in axes.flat:
        ax.set_ylim(common_ylim)
    for ax in axes.flat:
        ax.yaxis.set_major_locator(MaxNLocator(nbins=2, integer=True))
    dis_path = out_dir / "disentangle.png"
    fig.savefig(dis_path, bbox_inches=None, pad_inches=0.0, dpi=300)
    fig.savefig(out_dir / "disentangle.pdf", bbox_inches=None, pad_inches=0.0)
    plt.close(fig)
    # Optional norm summary plot
    fig_norm, ax_norm = plt.subplots(1, 1, figsize=(3, 2.5))
    ax_norm.plot(norm_reliable_list, label="Decay", marker="o", linewidth=2, color="#FF99CC", markersize=8)
    ax_norm.plot(norm_noise_list, label="Spindle", marker="^", linewidth=2, color="#006400", markersize=8)
    ax_norm.plot(norm_res_list, label="Residual", marker="s", linewidth=2, color="gray", markersize=8)
    ax_norm.set_xticks([0, 1, 2, 3], ["A", "B", "C", "D"])
    ax_norm.set_yticklabels([])
    ax_norm.grid(True, axis="y", alpha=0.22, linestyle="--", linewidth=0.45, color="#BBBBBB")
    fig_norm.savefig(out_dir / "norms.png", bbox_inches="tight", pad_inches=0.02, dpi=300)
    plt.close(fig_norm)
    return loss_path, dis_path
# ----------------------------- combined figure driver -------------------------
def plot_combined_model_figure(
    config: dict,
    output_root: Path,
    cache_root: Path,
    font_size: int = 8,
    show_legend: bool = True,
):
    """Generate a single figure with loss curve above the 4x4 distribution grid."""
    key = config["key"]
    out_dir = output_root / key
    out_dir.mkdir(parents=True, exist_ok=True)
    cache_dir = cache_root / key
    cache_dir.mkdir(parents=True, exist_ok=True)
    apply_style(font_size)
    epoch, train_loss, test_loss, loss_gap = _read_learning_csv(config["learning_csv"])
    timepoint_labels = config.get("timepoint_labels", TIMEPOINT_LABELS)
    if config.get("archived_dynamic", False):
        star_epoch = timepoint_labels[-1]
        star_interaction_dir = config["interaction_path"] / f"Harsanyi_{star_epoch}"
    else:
        # Star interaction reference at epoch 256
        star_epoch = 256
        star_interaction_dir = config["interaction_path"] / f"before_epoch_{star_epoch}"
    star_samples = sorted(os.listdir(star_interaction_dir))
    vn_v0_list = []
    for sample in star_samples:
        rewards = np.load(star_interaction_dir / sample / "before_sparsify" / "rewards.npy")
        vn_v0_list.append(np.abs(rewards[-1] - rewards[0]))
    salient_threshold_star = np.mean(vn_v0_list) * TAU
    star_I_and_list = []
    star_I_or_list = []
    masks = None
    for sample in star_samples:
        single_I_and, single_I_or, masks = _load_sample_interactions(
            star_interaction_dir / sample, salient_threshold_star
        )
        star_I_and_list.append(single_I_and)
        star_I_or_list.append(single_I_or)
    star_I_and_list = np.array(star_I_and_list)
    star_I_or_list = np.array(star_I_or_list)
    # Create combined figure: title + loss scaffold + timepoint strip + distribution grid.
    bar_row_heights = [h * DIS_SUBPLOT_HEIGHT for h in DIS_HEIGHT_RATIOS]
    title_height = COMBINED_TITLE_HEIGHT if show_legend else COMBINED_TITLE_HEIGHT + 0.10
    panel_figheight = COMBINED_FIGSIZE[1] if show_legend else COMBINED_FIGSIZE[1] - COMBINED_LEGEND_HEIGHT + 0.10
    fig = plt.figure(figsize=(COMBINED_FIGSIZE[0], panel_figheight))
    gs = fig.add_gridspec(
        5, 1,
        height_ratios=[
            title_height,
            COMBINED_LEGEND_HEIGHT if show_legend else 0.0,
            COMBINED_TOP_HEIGHT,
            COMBINED_TIMEPOINT_BAND_HEIGHT,
            BAR_GRID_HEIGHT,
        ],
        hspace=COMBINED_SECTION_HSPACE,
        left=COMBINED_LEFT,
        right=COMBINED_RIGHT,
        top=1.0 - COMBINED_TOP_MARGIN / panel_figheight,
        bottom=COMBINED_BOTTOM_MARGIN / panel_figheight,
    )
    gs_dist = gs[4, 0].subgridspec(
        4, 4,
        height_ratios=bar_row_heights,
        width_ratios=[1, 1, 1, 1],
        wspace=DIS_WSPACE,
        hspace=COMBINED_HSPACE,
    )
    # -------------------------------------------------------------------------
    # Row 0: panel title
    # -------------------------------------------------------------------------
    ax_title = fig.add_subplot(gs[0, 0])
    ax_title.axis("off")
    # Center over the plot content area: ax_title spans COMBINED_LEFT..COMBINED_RIGHT
    title_x = 0.5
    ax_title.text(
        title_x, 0.5, config["title"],
        ha="center", va="center",
        fontsize=9, fontweight="bold", color="#333333",
        transform=ax_title.transAxes,
    )
    # -------------------------------------------------------------------------
    # Row 1: dedicated horizontal legend strip
    # -------------------------------------------------------------------------
    ax_leg = fig.add_subplot(gs[1, 0]) if show_legend else None
    # -------------------------------------------------------------------------
    # Row 2: loss panel
    # -------------------------------------------------------------------------
    ax_loss = fig.add_subplot(gs[2, 0])
    _, remapped_timepoints, epoch_numbers = plot_loss_and_gap(
        epoch,
        train_loss,
        test_loss,
        loss_gap,
        config["loss_choose_index"],
        config["interact_choose_index"],
        config["interact_choose_index_in_loss_index"],
        ax=ax_loss,
        show_xaxis=False,
        legend_ax=ax_leg,
        show_legend=show_legend,
    )
    # -------------------------------------------------------------------------
    # Row 3: thin timepoint label strip
    # -------------------------------------------------------------------------
    ax_tp = fig.add_subplot(gs[3, 0])
    ax_tp.set_xlim(0, 4)
    ax_tp.set_ylim(0, 1)
    ax_tp.set_zorder(10)
    ax_tp.axis("off")
    label_y = 0.836
    epoch_label_y = 0.462
    for c, tp_label in enumerate(TIMEPOINT_LABELS):
        x = 0.5 + c
        ax_tp.text(
            x, label_y, f"Timepoint {tp_label}",
            ha="center", va="center",
            fontsize=FONT_SIZE_TIMEPOINT,
            color=TIMEPOINT_COLOR,
            fontweight="bold",
            transform=ax_tp.transData,
        )
        ax_tp.text(
            x, epoch_label_y, f"Epoch {epoch_numbers[remapped_timepoints[c]]}",
            ha="center", va="center",
            fontsize=6.5,
            color="#777777",
            transform=ax_tp.transData,
        )
    # Two-color phase-guidance arrow below timepoint labels (same style as top loss plot)
    arrow_y = 0.105
    arrow_lw = 2.0
    arrow_alpha = 0.85
    arrow_mutation = 8
    marker_size_normal = 3.5
    marker_size_transition = 5.0
    target_tp_x = np.array([0.5, 1.5, 2.5, 3.5])
    # A -> B: normal learning phase (line segment only)
    ax_tp.plot(
        [target_tp_x[0], target_tp_x[1]], [arrow_y, arrow_y],
        color=PHASE_NORMAL_ARROW_END,
        linewidth=arrow_lw,
        alpha=arrow_alpha,
        solid_capstyle="round",
        zorder=8,
    )
    # B -> D: over-fitting phase (line + arrowhead)
    ax_tp.annotate(
        "",
        xy=(target_tp_x[3], arrow_y),
        xytext=(target_tp_x[1], arrow_y),
        arrowprops=dict(
            arrowstyle="-|>",
            color=PHASE_OVERFIT_ARROW_END,
            lw=arrow_lw,
            alpha=arrow_alpha,
            mutation_scale=arrow_mutation,
            shrinkA=0,
            shrinkB=0,
        ),
        zorder=8,
    )
    # Hollow/lightly filled circular markers at A/B/C/D
    for i, x in enumerate(target_tp_x):
        is_b = i == 1
        size = marker_size_transition if is_b else marker_size_normal
        edge_color = PHASE_NORMAL_ARROW_END if i <= 1 else PHASE_OVERFIT_ARROW_END
        ax_tp.plot(
            x, arrow_y, "o",
            markersize=size,
            markerfacecolor="white",
            markeredgecolor=edge_color,
            markeredgewidth=1.5 if is_b else 1.0,
            alpha=arrow_alpha,
            zorder=9,
        )
    # -------------------------------------------------------------------------
    # Row 4: 4x4 distribution axes (no shared x so every panel keeps ticks)
    # -------------------------------------------------------------------------
    axes = np.empty((4, 4), dtype=object)
    for r in range(4):
        for c in range(4):
            kwargs = {}
            if c > 0:
                kwargs["sharey"] = axes[r, 0]
            axes[r, c] = fig.add_subplot(gs_dist[r, c], **kwargs)
    row_colors = ROW_COLORS
    norm_reliable_list = []
    norm_noise_list = []
    norm_res_list = []
    for fig_i, index in enumerate(config["interact_choose_index"]):
        if config.get("archived_dynamic", False):
            interaction_dir = config["interaction_path"] / f"Harsanyi_{timepoint_labels[fig_i]}"
        else:
            interaction_dir = config["interaction_path"] / f"before_{epoch[index]}"
        samples = sorted(os.listdir(interaction_dir))
        vn_v0_list = []
        for sample in samples:
            rewards = np.load(interaction_dir / sample / "before_sparsify" / "rewards.npy")
            vn_v0_list.append(np.abs(rewards[-1] - rewards[0]))
        salient_threshold = np.mean(vn_v0_list) * TAU
        single_I_and_list = []
        single_I_or_list = []
        for sample in samples:
            single_I_and, single_I_or, _ = _load_sample_interactions(
                interaction_dir / sample, salient_threshold
            )
            single_I_and_list.append(single_I_and)
            single_I_or_list.append(single_I_or)
        single_I_and_list = np.array(single_I_and_list)
        single_I_or_list = np.array(single_I_or_list)
        ax_col = axes[:, fig_i]
        cache_file = cache_dir / f"star_{star_epoch}"
        alpha, beta, sigma, gamma, norm_reliable, norm_noise, norm_res = disentangling_contributions(
            single_I_and_list,
            star_I_and_list,
            single_I_or_list,
            star_I_or_list,
            masks,
            ax_col,
            cache_file=cache_file,
            fig_i=fig_i,
            row_colors=row_colors,
        )
        norm_sum = norm_reliable + norm_noise + norm_res
        norm_reliable_list.append(norm_reliable / norm_sum)
        norm_noise_list.append(norm_noise / norm_sum)
        norm_res_list.append(norm_res / norm_sum)
        print(
            f"[{key}] combined epoch={epoch[index]} alpha={alpha:.4f} beta={beta:.4f} "
            f"sigma={sigma:.4f} gamma={gamma:.4f} reliable={norm_reliable_list[-1]:.4f} "
            f"noise={norm_noise_list[-1]:.4f} res={norm_res_list[-1]:.4f}"
        )
    # Unify all distribution axes to the first row's y-axis range
    ylim_mins = [axes[0, c].get_ylim()[0] for c in range(axes.shape[1])]
    ylim_maxs = [axes[0, c].get_ylim()[1] for c in range(axes.shape[1])]
    common_ylim = (min(ylim_mins), max(ylim_maxs))
    for ax in axes.flat:
        ax.set_ylim(common_ylim)
    for ax in axes.flat:
        ax.yaxis.set_major_locator(MaxNLocator(nbins=2, integer=True))
    if "dist_yticks" in config:
        for ax in axes.flat:
            ax.set_yticks(config["dist_yticks"])
    # Keep y-axis tick values only in the leftmost column of the 4x4 distribution grid.
    for r in range(axes.shape[0]):
        for c in range(1, axes.shape[1]):
            axes[r, c].tick_params(axis="y", labelleft=False, left=False)
    # --- Row background bands for vertical decomposition narrative ---
    # Row 0: warm background for "All".
    # Rows 1-3: cool background for decomposition components.
    row_bg_colors = ["#FBF0E8", "#F1F5FA", "#F1F5FA", "#F1F5FA"]
    row_bg_alphas = [0.60, 0.65, 0.65, 0.65]
    for r in range(4):
        for c in range(4):
            rect = plt.Rectangle(
                (0, 0), 1, 1,
                transform=axes[r, c].transAxes,
                facecolor=row_bg_colors[r],
                alpha=row_bg_alphas[r],
                edgecolor="none",
                zorder=-3,
            )
            axes[r, c].add_patch(rect)
    # --- Distribution row y-axis labels ---
    for r, label in enumerate(ROW_LABELS):
        axes[r, 0].set_ylabel(label, fontsize=FONT_SIZE_ROW_LABEL, color=ROW_LABEL_COLORS[r], labelpad=2)
    # --- Distribution block super-label ---
    dist_top = axes[0, 0].get_position().y1
    dist_bottom = axes[-1, 0].get_position().y0
    dist_center = (dist_top + dist_bottom) / 2
    row_label_x = 0.045 / COMBINED_FIGSIZE[0]  # 0.045 in from left edge (width-independent)
    fig.text(
        row_label_x,
        dist_center,
        "Interaction distributions",
        rotation=90,
        va="center",
        ha="center",
        fontsize=FONT_SIZE_ROW_LABEL + 2,
        color="#333333",
        fontweight="bold",
    )
    # --- Subtle orange background behind the Non-generalizable row ---
    for c in range(4):
        rect = plt.Rectangle(
            (0, 0), 1, 1,
            transform=axes[2, c].transAxes,
            facecolor=SPINDLE_BG_COLOR,
            alpha=SPINDLE_BG_ALPHA * 0.5,
            edgecolor="none",
            zorder=-1,
        )
        axes[2, c].add_patch(rect)
    combined_png_path = out_dir / "combined.png"
    combined_pdf_path = out_dir / "combined.pdf"
    fig.savefig(combined_png_path, bbox_inches=None, pad_inches=0.0, dpi=300)
    fig.savefig(combined_pdf_path, bbox_inches=None, pad_inches=0.0)
    plt.close(fig)
    return combined_png_path
# ----------------------------- individual exports -----------------------------
def export_individual_figures(
    source_root: Path = Path("source/dynamics"),
    output_dir: Path = Path("saved-results"),
    include_combined: bool = True,
):
    """Copy each generated source figure to the output directory with clear names."""
    source_root = Path(source_root)
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    for key in [cfg["key"] for cfg in MODELS]:
        sub = source_root / key
        exports = [("loss", "loss_and_gap.png"), ("loss", "loss_and_gap.pdf"), ("distribution", "disentangle.png"), ("distribution", "disentangle.pdf")]
        if include_combined:
            exports.extend([("combined", "combined.png"), ("combined", "combined.pdf")])
        for suffix, src_name in exports:
            src = sub / src_name
            dst = output_dir / f"{key}_{suffix}{Path(src_name).suffix}"
            if src.exists():
                shutil.copy2(src, dst)
                print(f"saved {dst}")
            else:
                print(f"missing {src}")
def export_combined_figures(
    source_root: Path = Path("source/dynamics"),
    output_dir: Path = Path("saved-results"),
    models=None,
):
    """Copy per-model combined PNG/PDF files to saved-results as {model_key}_combined.*."""
    source_root = Path(source_root)
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    if models is None:
        models = MODELS
    for key in [cfg["key"] for cfg in models]:
        for ext in ["png", "pdf"]:
            src = source_root / key / f"combined.{ext}"
            dst = output_dir / f"{key}_combined.{ext}"
            if src.exists():
                shutil.copy2(src, dst)
                print(f"saved {dst}")
            else:
                print(f"missing {src}")
# ----------------------------- combined composite -----------------------------
def composite_combined_figures(
    source_root: Path = Path("source/dynamics"),
    output_dir: Path = Path("saved-results"),
    models=None,
):
    """Compose per-model combined PNGs into compact PNG/PDF outputs."""
    if models is None:
        models = [
            ("AlexNet on CUB200-2011", "cub2011_alexnet"),
            ("Bert-Medium on SST-2", "sst2-bert"),
            ("VGG-16 on CUB200-2011", "cub2011_vgg16"),
            ("VGG-11 on CIFAR-10", "cifar10_vgg11"),
        ]
    source_root = Path(source_root)
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    panel_gap_x = GAP_X
    panel_gap_y = GAP_Y
    margin_x = 0
    margin_y = 0
    # No cropping: panels are already tightly laid out by plot_combined_model_figure.
    trim_top_px = 0
    trim_bottom_px = 0
    panel_meta = []
    for title, key in models:
        png_path = source_root / key / "combined.png"
        image = Image.open(png_path).convert("RGB")
        scale = PANEL_WIDTH / image.width
        panel_h = int(round(image.height * scale))
        if image.width != PANEL_WIDTH or image.height != panel_h:
            resample_filter = getattr(getattr(Image, "Resampling", Image), "LANCZOS")
            image = image.resize((PANEL_WIDTH, panel_h), resample_filter)
        # Cropping disabled: trim_top_px/trim_bottom_px are zero.
        panel_meta.append({"title": title, "key": key, "image": image, "height": panel_h})
    n_cols = 2
    n_rows = int(np.ceil(len(panel_meta) / n_cols))
    row_heights = [0] * n_rows
    for idx, meta in enumerate(panel_meta):
        row = idx // n_cols
        row_heights[row] = max(row_heights[row], meta["height"])
    fig_width = n_cols * PANEL_WIDTH + (n_cols - 1) * panel_gap_x + 2 * margin_x
    fig_height = sum(row_heights) + max(0, n_rows - 1) * panel_gap_y + 2 * margin_y
    canvas = Image.new("RGB", (int(fig_width), int(fig_height)), "white")
    row_offsets = [margin_y]
    for row in range(1, n_rows):
        row_offsets.append(row_offsets[-1] + row_heights[row - 1] + panel_gap_y)
    for idx, meta in enumerate(panel_meta):
        row, col = divmod(idx, n_cols)
        x0 = margin_x + col * (PANEL_WIDTH + panel_gap_x)
        y0 = row_offsets[row]
        canvas.paste(meta["image"], (int(x0), int(y0)))
    png_path = output_dir / "main_combined.png"
    canvas.save(png_path)
    print(f"saved {png_path}")
    # -------------------------------------------------------------------------
    # Vector PDF composition (from per-model combined.pdf)
    # -------------------------------------------------------------------------
    DPI = 300
    panel_width_pts = PANEL_WIDTH / DPI * 72
    panel_meta_pdf = []
    for title, key in models:
        pdf_path = source_root / key / "combined.pdf"
        reader = PdfReader(str(pdf_path))
        page = reader.pages[0]
        w = float(page.mediabox.width)
        h = float(page.mediabox.height)
        scale = panel_width_pts / w
        panel_meta_pdf.append({"page": page, "scale": scale, "height": h * scale})
    n_rows_pdf = int(np.ceil(len(panel_meta_pdf) / n_cols))
    row_heights_pdf = [0] * n_rows_pdf
    for idx, meta in enumerate(panel_meta_pdf):
        row = idx // n_cols
        row_heights_pdf[row] = max(row_heights_pdf[row], meta["height"])
    panel_gap_x_pts = panel_gap_x / DPI * 72
    panel_gap_y_pts = panel_gap_y / DPI * 72
    fig_width_pdf = n_cols * panel_width_pts + (n_cols - 1) * panel_gap_x_pts + 2 * margin_x
    fig_height_pdf = sum(row_heights_pdf) + max(0, n_rows_pdf - 1) * panel_gap_y_pts + 2 * margin_y
    writer = PdfWriter()
    new_page = writer.add_blank_page(width=fig_width_pdf, height=fig_height_pdf)
    for idx, meta in enumerate(panel_meta_pdf):
        row, col = divmod(idx, n_cols)
        x0 = margin_x + col * (panel_width_pts + panel_gap_x_pts)
        y0 = fig_height_pdf - margin_y - sum(row_heights_pdf[:row + 1]) - row * panel_gap_y_pts
        # Use an explicit CTM: translation must remain in output-page coordinates.
        scale = meta["scale"]
        transform = Transformation((scale, 0, 0, scale, x0, y0))
        new_page.merge_transformed_page(meta["page"], transform)
    pdf_path = output_dir / "main_combined.pdf"
    with open(pdf_path, "wb") as f:
        writer.write(f)
    print(f"saved {pdf_path}")
# ----------------------------- full pipeline ---------------------------------
def generate_all_source_figures(output_root: Path = Path("source/dynamics"), cache_root: Path = Path(".cache")):
    """Generate combined panels for all configured models."""
    output_root = Path(output_root)
    cache_root = Path(cache_root)
    for config in MODELS:
        plot_combined_model_figure(config, output_root, cache_root, FONT_SIZE)
    print(f"All combined panels saved to {output_root}")
# ----------------------------- deprecated composite -----------------------------
def composite_figures(*args, **kwargs):
    """Deprecated: the figure pipeline now writes PNG/PDF via composite_combined_figures()."""
    raise RuntimeError("Use composite_combined_figures(); legacy export is disabled.")


```
